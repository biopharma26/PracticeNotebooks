# Lab Session 4: Agricultural Genomics and Genome Editing
Design CRISPR-Cas9 guide RNAs and analyze target sequences for plant genome editing applications.

### 1. Environment Setup
Setting up string manipulation utilities for DNA sequence analysis.

```python
import re
print('Agricultural Genomics and CRISPR gRNA Design environment ready.')
```

### 2. Target Gene Sequence
Defining a target gene sequence for CRISPR-Cas9 guide RNA design.

```python
# Target gene sequence (e.g., in a crop plant like rice or tomato)
target_seq = "ATGGCCGATCGATCGATCGATCCGGATCGATCGATCGATCCGGATCGATCGATC"
print(f'Target Sequence Length: {len(target_seq)} bp')
```

### 3. CRISPR-Cas9 gRNA Finder (SpCas9 - NGG PAM)
Algorithmic scanning of target sequence to identify valid 20-bp protospacers adjacent to the 3'-NGG PAM site.

```python
def find_grnas(seq):
    # SpCas9 PAM is NGG. Target 20bp spacer upstream of PAM.
    # Pattern: 20 nucleotides followed by NGG
    pattern = re.compile(r'(?=(?P<spacer>[ACGT]{20})(?P<pam>[ACGT]GG))')
    matches = [m.group('spacer') + m.group('pam') for m in pattern.finditer(seq)]
    return matches

grnas = find_grnas(target_seq)
print(f'Found {len(grnas)} candidate gRNAs:')
for i, g in enumerate(grnas, 1):
    print(f'gRNA {i}: {g} (Spacer: {g[:-3]}, PAM: {g[-3:]})')
```
