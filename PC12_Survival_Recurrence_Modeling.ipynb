```python
from lifelines import KaplanMeierFitter, CoxPHFitter
n=120
df=pd.DataFrame({
    "time":np.random.exponential(900,n).astype(int)+1,
    "event":np.random.binomial(1,.45,n),
    "risk_signature":np.random.normal(size=n),
    "age":np.random.normal(65,7,n),
    "psa":np.random.lognormal(2.2,.7,n)
})
```

```python
km=KaplanMeierFitter()
median=df.risk_signature.median()
for label,mask in [("Low",df.risk_signature<=median),("High",df.risk_signature>median)]:
    km.fit(df.loc[mask,"time"],df.loc[mask,"event"],label=label)
    km.plot_survival_function()
plt.title("Kaplan–Meier demonstration"); plt.xlabel("Days"); plt.ylabel("Survival probability"); plt.show()
```

```python
cph=CoxPHFitter()
cph.fit(df,"time",event_col="event")
cph.print_summary()
```
