# DNA Methylation Analysis

## Introduction

Study methylation beta value distributions.

## Load Dataset



```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'beta_value': np.random.rand(1000)
})
data.head()
```

## Visualization



```python
import matplotlib.pyplot as plt

plt.hist(data['beta_value'], bins=30)
plt.title("Methylation Distribution")
plt.xlabel("Beta value")
plt.ylabel("Frequency")
plt.show()
```
