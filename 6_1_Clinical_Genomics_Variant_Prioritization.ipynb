This notebook guides MSc students through clinical variant annotation filtering and prioritization using Python.

### 1. Environment Setup & Dependencies
Install and import standard libraries for handling tabular genomic variant data.

```python
import pandas as pd
import numpy as np

print('Libraries imported successfully for Clinical Variant Prioritization.')
```

### 2. Simulate Clinical VCF / Variant Data
We create a mock dataframe representing annotated variant calls from an exome sequencing run.

```python
data = {
    'Gene': ['BRCA1', 'TP53', 'EGFR', 'CFTR', 'TTN'],
    'Variant': ['c.68_69delAG', 'c.743G>A', 'c.2573T>G', 'c.1521_1523delCTT', 'c.1234A>T'],
    'gnomAD_MAF': [0.00001, 0.000, 0.002, 0.005, 0.045],
    'CADD_Phred': [34.5, 28.1, 24.3, 18.2, 5.1],
    'ClinVar': ['Pathogenic', 'Pathogenic', 'Pathogenic', 'Likely Benign', 'Benign']
}

df_variants = pd.DataFrame(data)
df_variants
```

### 3. Variant Filtering Pipeline
Applying strict filtering criteria based on population frequency and pathogenicity prediction scores.

```python
# Filter based on clinical significance thresholds: MAF < 0.01 and CADD > 20
filtered_variants = df_variants[(df_variants['gnomAD_MAF'] < 0.01) & (df_variants['CADD_Phred'] > 20)]
print('Prioritized Pathogenic/Likely Pathogenic Candidates:')
filtered_variants
```
