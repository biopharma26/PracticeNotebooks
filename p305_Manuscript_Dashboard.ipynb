# 05 — Manuscript Figure Assembly & Master Dashboard

Combines the outputs of Notebooks 01–04 into a single publication-style 2×2 figure panel, matching the
layout of the uploaded report. **Run Notebooks 01–04 first** (in the same Colab session, or upload their
saved `.png`/`.csv` outputs to this notebook's working directory) so the files referenced below exist.


```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

plt.rcParams["figure.dpi"] = 150

expected_files = [
    "figure2_deg_volcano.png",
    "figure1_km_pvs.png",
    "figure3_admet_heuristic.png",
    "figure4_md_dashboard_PLACEHOLDER.png",
]
missing = [f for f in expected_files if not os.path.exists(f)]
if missing:
    print("Missing (run the corresponding notebook first, or upload the file):", missing)
else:
    print("All figures found.")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 11))
titles = [
    "A. Transcriptomics: DEG Volcano Plot",
    "B. Multi-Omics PVS: Kaplan-Meier Stratification",
    "C. CADD: ADMET Liability Heuristic",
    "D. MD Trajectory Dashboard (placeholder until real MD is run)",
]

for ax, fname, title in zip(axes.flat, expected_files, titles):
    ax.set_title(title, fontsize=11, loc="left")
    ax.axis("off")
    if os.path.exists(fname):
        img = mpimg.imread(fname)
        ax.imshow(img)
    else:
        ax.text(0.5, 0.5, f"[{fname} not found —\nrun the source notebook]",
                ha="center", va="center", fontsize=9, color="firebrick")

fig.suptitle("Precision Oncology Mechanisms of Pacific Yew Taxanes — Master Dashboard",
             fontsize=14, y=1.00)
plt.tight_layout()
plt.savefig("master_dashboard.png", dpi=300, bbox_inches="tight")
plt.show()
```

## Manuscript-readiness checklist

Before this pipeline's output goes into a manuscript or dissertation chapter, confirm each item —
these are exactly the substitutions flagged in Notebooks 01–04:

| Component | Current status in this pipeline | Required before publication |
|---|---|---|
| DEG / volcano plot | Synthetic count matrix | Real TCGA/GEO counts + DESeq2/pydeseq2 |
| PVS score & survival | Simulated 500-patient cohort, Cox-fitted weights on synthetic data | Real clinical + multi-omics cohort, cross-validated weights |
| Molecular descriptors | **Real** — RDKit-computed from verified structures | Already publication-ready |
| ADMET (Pgp/BBB) | Rule-based heuristic from real descriptors | Cited experimental assay data or validated ADMET tool |
| Docking ΔG | Scaffold only, not executed | Real AutoDock Vina run vs. a deposited tubulin structure |
| MD (RMSD/RMSF/Rg) | Scaffold + illustrative placeholder | Real OpenMM/GROMACS trajectory + MDAnalysis output |

here is what makes the pipeline defensible to reviewers.
